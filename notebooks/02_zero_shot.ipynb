```python
import sys, os
from pathlib import Path
# Make src/ importable whether run from repo root or notebooks/.
ROOT = Path.cwd()
if (ROOT / 'src').exists():
    sys.path.insert(0, str(ROOT / 'src'))
elif (ROOT.parent / 'src').exists():
    ROOT = ROOT.parent
    sys.path.insert(0, str(ROOT / 'src'))
os.chdir(ROOT)
print('project root:', ROOT)
```

```python
import data_loader as dl, srl_inference as si, evaluation as ev
client = si.get_client()
MODEL = si.DEFAULT_MODEL
print('model:', MODEL)
```

## Sanity check on a few examples first
Do not run the whole set before confirming the pipeline works.

```python
test = dl.read_jsonl('data/test.jsonl')
sample = test[:10]
results = si.predict_batch(client, sample, model=MODEL)
for r in results[:3]:
    print(r['sentence'])
    print('  ->', r['prediction']['arguments'])
    print('  schema_valid:', r['schema_valid'])
```

## Evaluate the sample

```python
m = ev.evaluate(results, sample)
ev.summary_row('zero-shot (sample)', m)
```

## Full zero-shot run
Runs over the whole test set and saves predictions for later comparison.

```python
results = si.predict_batch(client, test, model=MODEL)
si.save_predictions(results, 'results/predictions_zero_shot.jsonl')
m = ev.evaluate(results, test)
ev.summary_row('zero-shot', m)
```
